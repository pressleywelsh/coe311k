```python
def leastSq(X,Y):
    n = len(X)
    sumX = sum(X)
    sumY = sum(Y)
    sumX2 = sum(X**2)
    sumXY = sum(X*Y)
    m = ((n * sumXY) - (sumX * sumY))/((n * sumX2) - (sumX**2))
    c = (sumY - (m * sumX))/n
    return m,c
```
