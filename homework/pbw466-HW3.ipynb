```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
def approxSin(ind, x): 
    approx=0.0
    for i in range(0,ind):
        approx += (((-1)**i) * ((x**((2*i)+1)) / math.factorial((2*i)+1)))
    return approx
```

```python
for n in range(1,5):
    x = []
    y = []
    for i in np.arange(0,10,0.1):
        calcSin = approxSin(n, i)
        error = abs(np.sin(i) - calcSin)
        x.append(i)
        y.append(error)
    plt.plot(x,y, label = f"{n} terms")
plt.xlabel("x")
plt.ylabel("Error")
plt.legend()
plt.show()
```

## The truncation error gets smaller as you increase the amount of terms only when you are using small values of x. At small values of x, increasing number of terms decreases error. The sin(x) taylor series relies on the behavior and derivatives at 0, making it more accurate when x is closer to 0. For large values of x, the taylor series is not an accurate depiction of sin(x). My graph shows that at the further end of the x-axis, using 4 terms results in more error.
