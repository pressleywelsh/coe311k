```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi
N = 50

x = np.linspace(a, b, N)

h = (b - a) / (N - 1)

exact = np.cos(x)

central = (f(x + h) - f(x - h)) / (2 * h)

forward = (f(x + h) - f(x)) / h

backward = (f(x) - f(x - h)) / h
```

```python
plt.figure()

plt.subplot(2, 2, 1)
plt.plot(x, exact)
plt.title("Exact Derivative: cos(x)")

plt.subplot(2, 2, 2)
plt.plot(x, forward)
plt.title("Forward Difference")

plt.subplot(2, 2, 3)
plt.plot(x, backward)
plt.title("Backward Difference")

plt.subplot(2, 2, 4)
plt.plot(x, central)
plt.title("Central Difference")

plt.show()
```
