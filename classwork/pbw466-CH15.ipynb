```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

x_start, x_end = 0, 2

first_run_y = None
first_run_x = None

plt.figure()

for h in np.arange(0.01, 1.0, 0.05):

    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)

    y_values[0] = 1

    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    plt.plot(x_values, y_values, label=f"h={h}")

    if first_run_y is None:
        first_run_y = y_values
        first_run_x = x_values

        print(f"Reference run h={h}")
        print("Pointwise error = 0")
        print("Total accumulated error = 0")

    else:
        y_interp = np.interp(first_run_x, x_values, y_values)
        error = np.abs(first_run_y - y_interp)
        total_error = np.sum(error)

        print(f"\nRun h={h:.2f}")
        print("Pointwise error compared to first run:")
        print(error)
        print("Total accumulated error:")
        print(total_error)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler's Method with Different Step Sizes")
plt.legend()
plt.show()
```
