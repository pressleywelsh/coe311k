```python
import numpy as np

import matplotlib.pyplot as plt

from scipy.interpolate import CubicSpline

 

# Define known data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

# Create a cubic spline interpolation of the data
cs = CubicSpline(x, y, bc_type='natural')

# Create a polyfit
coeffs = np.polyfit(x, y, 3)
poly = np.poly1d(coeffs)

# Generate new x values for a smooth plot
x_interp = np.linspace(0, 5, 100)
y_interp = cs(x_interp)
y_spline = cs(x_interp)
y_poly = poly(x_interp)
```

```python
# Plot original data and cubic spline interpolation
plt.plot(x, y, 'bo', label="Data points")
plt.plot(x_interp, y_interp, 'r-', label="Cubic Spline Interpolation")
plt.title("Cubic Spline Interpolation")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
plt.scatter(x, y, label='Data Points')
plt.plot(x_interp, y_spline, label='Cubic Spline')
plt.plot(x_interp, y_poly, label='Polyfit (degree 3)')
plt.legend()
plt.grid(True)
plt.show()
```
