```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters

k = 10.0        # Constant in the equation (e.g., spring constant)
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]       # Step size
x_start = 0    # Start of the interval
x_end = 10     # End of the interval
y_0 = 1.0      # Initial condition for y
v_0 = 0.0      # Initial condition for dy/dx

# Euler's Method loop
plt.figure(figsize=(12, 6))

for h in h_values:
    # Create arrays
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    # Initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    # Euler's Method loop 
    for i in range(1, len(x_values)):
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]    

        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

    # Plot position and velocity
    plt.plot(x_values, y_values, label=f'y (h={h})')
    plt.plot(x_values, v_values, linestyle='--', label=f'v (h={h})')

plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method (different h)')
plt.legend()
plt.grid(True)

plt.show()
```

As the h values increase, the graph becomes more unstable. It starts to become noticeable when h hits 0.1 and becomes increasingly less stable from there. I noticed the graph appears to oscillate faster as the k-value increases. 
When k=10.0 and h=0.01, the graph is more stable than larger h-values. The graph is still less stable than when k=1.0. 
I believe that large k-values just have this effect on all step sizes. It is just a more noticeable effect on large h-values. Euler's method is not able to solve this with a larger h because the solution starts to become unstable.
