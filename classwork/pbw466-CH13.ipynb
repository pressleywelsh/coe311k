```python
def matrix_mult(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    
    return C
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time

sizes = [3]
for i in range(7):
    sizes.append(sizes[-1] * 2)

naive_times = []
numpy_times = []
matrix_numbers = [1, 2, 3, 4, 5, 6, 7, 8]

for n in sizes:
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)

    A_list = A.tolist()
    B_list = B.tolist()

    start = time.time()
    matrix_mult(A_list, B_list)
    end = time.time()
    naive_times.append(end - start)

    start = time.time()
    np.dot(A, B)
    end = time.time()
    numpy_times.append(end - start)
```

```python
plt.plot(matrix_numbers, naive_times, marker='o', label='Naive')
plt.plot(matrix_numbers, numpy_times, marker='o', label='NumPy')
plt.title('Matrix Multiplication Runtime Comparison')
plt.xlabel('Matrix Number')
plt.ylabel('Runtime (seconds)')
plt.legend()
plt.show()
```
