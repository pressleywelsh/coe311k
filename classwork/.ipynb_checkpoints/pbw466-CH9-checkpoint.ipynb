```python
import numpy as np

# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T
print("X: ")
print(X)

print("Y: ")
print(Y)

print("A: ")
print(A)
```

```python
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")
```

```python
import matplotlib.pyplot as plt

x_line = np.linspace(min(X), max(X), 100)
y_line = m*x_line + c

plt.scatter(X,Y, label = "Data Points")
plt.plot(x_line, y_line, label = "Best fit line: y = mx + c")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Least squares fit")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import random

X2 = np.random.rand(10)
X2 = np.sort(X2)
Y2 = np.random.rand(10)

A2 = np.vstack([X2, np.ones(len(X2))]).T
m2, c2 = np.linalg.lstsq(A2, Y2, rcond=None)[0]

print(f"Slope: {m2}, Intercept: {c2}")

x_line = np.linspace(min(X2), max(X2), 100)
y_line = m2 * x_line + c2

plt.scatter(X2, Y2, label="Random data")
plt.plot(x_line, y_line, color="red", label="Best fit line")

plt.legend()
plt.show()
```

```python
m_true = 2.5
b_true = -0.4

X3 = np.arange(0, 1.01, 0.01)
noise = np.random.uniform(-1, 1, len(X3))
Y3 = m_true*X3 + b_true + noise

A3 = np.vstack([X3, np.ones(len(X3))]).T
m3, b3 = np.linalg.lstsq(A3, Y3, rcond=None)[0]

print("True:", m_true, b_true)
print("Estimated:", m3, b3)

x_line = np.linspace(0, 1, 200)
plt.scatter(X3, Y3, s=10, label="Noisy data")
plt.plot(x_line, m3*x_line + b3, label="Least squares fit")
plt.plot(x_line, m_true*x_line + b_true, linestyle="--", label="True line")
plt.legend()
plt.show()
```
