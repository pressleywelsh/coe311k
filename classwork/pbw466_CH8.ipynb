```python
def dominant(mat):
  diagDom = True
  for i in range(len(mat)):
    sum = 0
    for j in range(len(mat)):
      if (i == j):
        diag = abs(mat[i, j])
      else:
        sum += abs(mat[i, j])
    if (sum>diag):
      return False
  return True
```
